## Using boundary box

The below example will be using the boundary box functionality. We will be generationg the height data for Kebnekaise, using the boundary box:

bbox = ($18.38155^\circ$ E, $67.88428^\circ$ N, $18.52011^\circ$ E, $67.90249^\circ$ N)

We will be defining 500 grid points along this tile.
These grid points are essentially potential test points for further use cases.

```python
# kebnekaise
region = Region.from_bounds(
    west=18.38155,
    east=18.52011,
    south=67.88428,
    north=67.90249
)

terrain = region.generate(nr_points=500, demtype="Lantmateriet", workers=1)
```

```python
terrain.plot_grid(nr_points = 1000)
```
